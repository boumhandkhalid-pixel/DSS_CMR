#Donnees indices

```python
xls = pd.ExcelFile('Compo_All_Indices_20260731_copy.xlsx')
sheet_names = xls.sheet_names
print(f"Sheets in the Excel file: {sheet_names}")
```

Now, you can read all sheets into a dictionary where each key is the sheet name and each value is a pandas DataFrame.

```python
all_sheets_df = pd.read_excel('Compo_All_Indices_20260731_copy.xlsx', sheet_name=None)

# To see the data for a specific sheet, you can access it like this:
# For example, to see the head of the first sheet:
for num, sheet_name in enumerate(sheet_names, 1):
    print(f"\nDisplaying the head of the {num} sheet ('{sheet_name}'):")
    display(all_sheets_df[sheet_name].head())
```
