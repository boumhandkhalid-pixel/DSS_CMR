## ⚠️  MANDATORY STEP BEFORE PRODUCTION

This notebook evaluates whether the BASELINE_HYPOTHESIS weights and thresholds
produce profitable trading decisions.

**What we evaluate:**
- Hit rate (% of BUY decisions where forward_return > 0)
- Risk-adjusted returns (Sharpe ratio)
- Maximum drawdown (worst peak-to-trough loss)
- Robustness across weight configurations A–E
- Stability across time periods (development/validation/test)

**What we DO NOT do:**
- Tune thresholds to maximize test-period returns (overfitting)
- Cherry-pick weight config based on test results
- Claim the strategy is profitable without statistical evidence

## Temporal Split (Strictly Chronological)

```
Development  60%  → Observe performance, compare configs
Validation   20%  → Confirm selected config stability
Test         20%  → Final evaluation (NEVER USED FOR TUNING)
```
