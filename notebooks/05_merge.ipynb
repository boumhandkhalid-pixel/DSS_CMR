# Notebook 05 — Independent Dataset Design

This notebook documents the recommended separation between the normalized market dataset and the index-composition dataset.

## Design rule

Do not merge both datasets into a single master table. Keep them independent and use the index-composition dataset only during the dynamic filtering stage.

```python
import pandas as pd

market_df = pd.DataFrame({
    'Date': ['2026-07-20', '2026-07-21'],
    'CODE_ISIN': ['MA0000012296', 'MA0000012296'],
    'Company': ['AFMA', 'AFMA'],
    'Close Price': [1230.0, 1231.0],
    'Volume MC': [2150000, 2340000],
})

index_df = pd.DataFrame({
    'Date': ['2026-07-21'],
    'CODE_ISIN': ['MA0000012296'],
    'Index': ['MASI'],
    'Weight': [0.12],
    'Free Float Factor': [0.80],
})

print('Market dataset shape:', market_df.shape)
print('Index composition shape:', index_df.shape)
print('\nThey remain independent datasets for different stages of the pipeline.')
```
